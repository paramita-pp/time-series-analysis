# Turning Bullish

Source: https://raw.githubusercontent.com/paramita-pp/time-series-analysis/refs/heads/main/Files/SPY_close_price_5Y.csv

Task:
- To calculate 50-day and 20-day moving averages based on the closing price
- Identify every "Golden Corss" moment - when the short-term average (50-day) crosses from below to above long-term average (200-day)


```python
# import library
import pandas as pd
```

```python
# import data
spy_df = pd.read_csv('https://raw.githubusercontent.com/paramita-pp/time-series-analysis/refs/heads/main/Files/SPY_close_price_5Y.csv', parse_dates  = ['Date'])

spy_df.head()
```

```python
spy_df.assign(
    ma_50 = lambda d: d.Close.rolling(window = 50, min_periods = 50).mean(), # moving average 50 days
    ma_200 = lambda d: d.Close.rolling(window = 200, min_periods = 200).mean(), # moving average 200 days
    gc = lambda d: ((d.ma_50 > d.ma_200) & (d.ma_50.shift(1) < d.ma_200.shift(1))).astype(int) # golden cross check
).query("gc==1") # query only golden cross
```
